```python
#selectioncontrol, if-elif statement

score = int(input("Enter the student's score:"))
if score >= 90:
  print("CHINESE MOM APPROVED")
elif score <= 75:
  print("U FAILURE GET TF OUT OF HERE")
```

```python
#selection control, match, case _ if

score = int(input("Enter the student's score:"))
match score:
  case score if score >= 90:
    print("CHINESE MOM APPROVED")
  case score if score <= 80:
    print("meh u can do better")
```

```python
#for iteration

for number in range(0, 40):
  if number % 5 == 0:
    print(number)

```

```python
#while iteration

password = "imnotafemboy"

max_tries = 5
attempts_made = 0

while True:
  entered_password = input("Enter password:")

  if entered_password == password:
    print("Access Granted!")

  else:
    attempts_made += 1
    print("Incorrect Password. TRY AGAIN.")

    if attempts_made == max_tries:
      print("Maximum attempts reached. ACCESS DENIED.")
      break

else:
    print(f"Attempts remaining: {max_tries - attempts_made}")

```

```python
#loopcontrolstatements

for number in range(1, 10):
  if number == 3:
    continue
  elif number == 6:
    pass
  elif number == 9:
    break
  print(number)
```

```python
#function
def calculate_total(price, quantity):
  total = price * quantity
  return total


#MAIN PROGRAM

try:
   price = float(input("Enter the price of the item:"))
   quantity = int(input("Enter the quantity of the item:"))

   total = calculate_total(price, quantity)
   print(f"\nPrice per item: ₱{price:.2f}")
   print(f"Quantity: {quantity}")
   print(f"Total: ₱{total:.2f}")

except ValueError:
   print("Invalid input. Please enter valid numbers for price and quantity.")
```
