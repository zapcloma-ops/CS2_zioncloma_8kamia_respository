```python
#selection control, if-else

num = int(input("Enter a number:"))

if num % 2 == 0:
  print("The number is even.")
else:
  print("The number is odd.")
```

```python
#selection control, match case_if

num = int(input("Enter student's NUMERICAL grade equivalent:"))
match num:
 case num if num >= 90:
   print("Student's grade is A.")
 case num if num >= 80:
   print("Student's grade is B.")
 case num if num >= 70:
   print("Student's grade is C.")
 case num if num >= 60:
   print("Student's grade is D.")
 case num if num >= 0:
   print("Student's grade is F.")

```

```python
#"for" iteration

for number in range(0, 100):
  if number % 10 == 0:
    print(number)
```

```python
#"while" iteration

password = "@6seven"

max_tries = 4
attempts_made = 0

while True:
  entered_password = input("Enter password.")
  if entered_password == password:
    print("Access Granted!")
  else:
    attempts_made += 1
    print("Incorrect Password. Invalid. TRY AGAIN")

    if attempts_made == max_tries:
      print("Maximum attempts reached. ACCESS DENIED.")
    break

else:
  print(f"Attempts Remaining: {max_tries - attempts_made}")
```

```python
#loop control statements

for num in range(1, 20):
  if number == 5:
    continue
  elif number == 10:
    pass
  elif number == 15:
    break
print(number)
```

```python
#function
def calculate_total(price, quantity):
  total = price * quantity
  return total

#MAIN PROGRAM
try:
  price = float(input("Enter product price:"))
  quantity = int(input("Enter product quantity:"))

  total = calculate_total(price, quantity)
  print = (f"\nPrice per item: ₱{price:.2f}")
  print = (f"Quantity: {quantity}")
  print = (f"Total: ₱{total:.2f}")

except ValueError:
    print("Invalid input. Please enter valid numbers.")
```
