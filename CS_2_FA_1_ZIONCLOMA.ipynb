```python
#item61
import math

x1 = float(input("Enter x1: "))
x2 = float(input("Enter x2: "))
y1 = float(input("Enter y1: "))
y2 = float(input("Enter y2: "))

distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))

print(f"Distance: {distance}")
```

```python
#item2
import math

r = float(input("Enter the radius: "))
area = math.pi * math.pow(r, 2)
circumference = math.pi * 2 * r

print(f"Area: {area}")
print(f"Circumference: {circumference}")
```

```python
#item3
import math

leg1 = float(input("Enter the first leg: "))
leg2 = float(input("Enter the second leg: "))

hypotenuse1 = math.hypot(leg1, leg2)
hypotenuse2 = math.sqrt(math.pow(leg1, 2) + math.pow(leg2, 2))

print(f"Hypotenuse (using math.hypot): {hypotenuse1}")
print(f"Hypotenuse (using math.sqrt and math.pow): {hypotenuse2}")
```
